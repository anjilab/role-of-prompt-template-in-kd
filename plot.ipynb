```python
import matplotlib.pyplot as plt
import numpy as np

# Data
dolly_safety = [26, 26.14, 26.74, 26.52, 27.2, 27.5, 27.42, 27.2, 27.57]
llama_safety = [26, 28.71, 30.45, 30.38, 30.3, 30.3, 31.06, 30.61, 30.46] 
dolly_utility = [18.77, 21.31, 22.35, 22.66, 22.89, 23.03, 23.27, 23.26, 23.25] 
llama_utility = [18.77, 26.62, 28.09, 28.81, 28.4, 28.59, 28.58, 28.61, 28.64]

steps = np.arange(len(dolly_safety))

# Global Font Size Settings
FONT_SIZE_LABEL = 20
FONT_SIZE_TICKS = 16
FONT_SIZE_LEGEND = 14 # Increased slightly for consistency

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6)) 

# --- Plot 1: Utility ---
ax1.plot(steps, dolly_utility, marker='o', markersize=8, linestyle='-', color='#1f77b4', linewidth=2.5, label='Non-chat')
ax1.plot(steps, llama_utility, marker='o', markersize=8, linestyle='-', color='#ff7f0e', linewidth=2.5, label='Chat')
ax1.set_xlabel('Distillation Steps', fontsize=FONT_SIZE_LABEL)
ax1.set_ylabel('DollyEval(R-L)', fontsize=FONT_SIZE_LABEL)
ax1.tick_params(axis='both', which='major', labelsize=FONT_SIZE_TICKS)
ax1.grid(True, linestyle='--', alpha=0.7)
ax1.legend(fontsize=FONT_SIZE_LEGEND, loc='lower right')

# --- Plot 2: Safety ---
ax2.plot(steps, dolly_safety, marker='o', markersize=8, linestyle='-', color='#1f77b4', linewidth=2.5, label='Non-chat')
ax2.plot(steps, llama_safety, marker='o', markersize=8, linestyle='-', color='#ff7f0e', linewidth=2.5, label='Chat')
ax2.set_xlabel('Distillation Steps', fontsize=FONT_SIZE_LABEL)
ax2.set_ylabel('SORRY-Bench (ASR%)', fontsize=FONT_SIZE_LABEL)
ax2.tick_params(axis='both', which='major', labelsize=FONT_SIZE_TICKS)
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.legend(fontsize=FONT_SIZE_LEGEND, loc='lower right')

# Increase the gap between plots
# wspace=0.3 provides a significant visible gap
plt.subplots_adjust(wspace=0.35)

# Use bbox_inches='tight' during save to capture everything correctly
plt.savefig("kd_prompt_utility_safety_comparison.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
